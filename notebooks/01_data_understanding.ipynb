```python
import os
import duckdb
import pandas as pd

con = duckdb.connect()
raw_dir = '../data/raw'
print('Available Raw CSV Files:', os.listdir(raw_dir))
```

### 1. Ingest and Profile Table Sizes


```python
tables = ['campaign_desc', 'campaign_table', 'coupon', 'coupon_redempt', 'hh_demographic', 'product', 'transaction_data', 'causal_data']
inventory = []
for t in tables:
    cnt = con.execute(f"SELECT COUNT(*) FROM read_csv_auto('{raw_dir}/{t}.csv')").fetchone()[0]
    inventory.append({'Table': t, 'Rows': cnt})
pd.DataFrame(inventory)
```
